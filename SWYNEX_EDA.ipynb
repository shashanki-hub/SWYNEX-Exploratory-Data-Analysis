```python

import matplotlib.pyplot as plt

eda_df["product_category"].value_counts().plot(
    kind="barh", figsize=(8, 5), color="steelblue"
)

plt.title("Product Category Distribution")
plt.xlabel("Number of Records")
plt.ylabel("Product Category")
plt.tight_layout()
plt.show()
```

```python

category_sales = (
    eda_df.groupby("product_category")["estimated_sales_inr"]
    .sum()
    .sort_values()
)

category_sales.plot(kind="barh", figsize=(8, 5), color="teal")

plt.title("Estimated Sales by Product Category")
plt.xlabel("Estimated Sales (INR)")
plt.ylabel("Product Category")
plt.tight_layout()
plt.show()
```

```python

brand_sales = (
    eda_df.groupby("brand")["estimated_sales_inr"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
    .sort_values()
)

brand_sales.plot(kind="barh", figsize=(8, 6), color="mediumpurple")

plt.title("Top 10 Brands by Estimated Sales")
plt.xlabel("Estimated Sales (INR)")
plt.ylabel("Brand")
plt.tight_layout()
plt.show()
```

```python

channel_counts = eda_df["sales_channel"].value_counts().dropna()

channel_counts.sort_values().plot(
    kind="barh", figsize=(8, 5), color="cornflowerblue"
)

plt.title("Records by Sales Channel")
plt.xlabel("Number of Records")
plt.ylabel("Sales Channel")
plt.tight_layout()
plt.show()
```

```python

return_counts = (
    eda_df["returned"]
    .astype("string")
    .str.strip()
    .str.title()
    .value_counts()
    .reindex(["Yes", "No"], fill_value=0)
)

return_counts.plot(
    kind="bar", figsize=(7, 5),
    color=["tomato", "mediumseagreen"]
)

plt.title("Returned vs Non-Returned Records")
plt.xlabel("Returned")
plt.ylabel("Number of Records")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
