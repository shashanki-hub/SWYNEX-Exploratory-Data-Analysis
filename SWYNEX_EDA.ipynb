```python

import pandas as pd
import numpy as np
from pathlib import Path

# ==================================================
# 1. LOAD THE ORIGINAL DATASET
# ==================================================
file_path = Path(
    r"D:\SWYNEX-Data-Cleaning-Preparation\DATA CLEANING\Unclean data.csv"
)

df = pd.read_csv(file_path, dtype=str, keep_default_na=False)
original_rows = len(df)

# Keep a backup in memory
original = df.copy(deep=True)

# ==================================================
# 2. STANDARDIZE COLUMN NAMES
# ==================================================
df.columns = (
    df.columns.str.strip()
    .str.lower()
    .str.replace(r"\s+", "_", regex=True)
    .str.replace(r"[^\w]", "", regex=True)
)

# ==================================================
# 3. CLEAN TEXT AND MISSING VALUES
# ==================================================
missing_markers = {
    "", "na", "n/a", "n.a.", "nan", "null",
    "none", "not applicable", "n.a", "-"
}

for col in df.columns:
    df[col] = df[col].str.strip()

    df[col] = df[col].apply(
        lambda x: pd.NA
        if pd.isna(x) or x.strip().lower() in missing_markers
        else x.strip()
    )

# ==================================================
# 4. CLEAN TRANSACTION ID (PRIMARY KEY)
# ==================================================
df["transaction_id"] = (
    df["transaction_id"]
    .str.upper()
    .str.replace(r"[\s-]+", "", regex=True)
)

df["transaction_id"] = df["transaction_id"].replace("", pd.NA)

# Do not invent missing IDs or silently delete duplicates
missing_ids = df["transaction_id"].isna().sum()
duplicate_ids = df["transaction_id"].duplicated(keep=False).sum()

print("Missing transaction IDs:", missing_ids)
print("Rows involved in duplicate IDs:", duplicate_ids)

# ==================================================
# 5. STANDARDIZE CATEGORICAL VALUES
# ==================================================
category_cols = [
    "product_category", "product_name", "brand",
    "skin_type", "sales_channel", "region",
    "stock_status", "payment_method"
]

for col in category_cols:
    if col in df.columns:
        df[col] = df[col].str.title()

# Correct known category variations
if "product_category" in df.columns:
    df["product_category"] = df["product_category"].replace({
        "Body Care!": "Body Care",
        "Fragrance!": "Fragrance"
    })

if "sales_channel" in df.columns:
    df["sales_channel"] = df["sales_channel"].replace({
        "Retail Store": "Retail Store",
        "Online Store": "Online Store",
        "Social Commerce": "Social Commerce",
        "Marketplace": "Marketplace"
    })

if "stock_status" in df.columns:
    df["stock_status"] = (
        df["stock_status"].str.lower().map({
            "in stock": "In Stock",
            "in_stock": "In Stock",
            "out of stock": "Out Of Stock",
            "out_of_stock": "Out Of Stock",
            "low stock": "Low Stock",
            "low_stock": "Low Stock"
        })
    )

if "returned" in df.columns:
    df["returned"] = df["returned"].str.lower().map({
        "yes": "Yes",
        "no": "No"
    })

# ==================================================
# 6. CONVERT DATA TYPES
# ==================================================
# Mixed date formats exist in this dataset.
# dayfirst=True is an assumption for ambiguous slash dates.
df["order_date"] = pd.to_datetime(
    df["order_date"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)

numeric_cols = [
    "unit_price_inr", "discount_percent",
    "quantity", "customer_age", "rating"
]

for col in numeric_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

# Nullable integers preserve missing values
for col in ["quantity", "customer_age"]:
    if col in df.columns:
        df[col] = df[col].round().astype("Int64")

# ==================================================
# 7. VALIDATE VALUES
# ==================================================
# Invalid values become missing; original data is backed up above.
if "unit_price_inr" in df.columns:
    df.loc[df["unit_price_inr"] <= 0, "unit_price_inr"] = np.nan

if "discount_percent" in df.columns:
    df.loc[
        ~df["discount_percent"].between(0, 100),
        "discount_percent"
    ] = np.nan

if "customer_age" in df.columns:
    df.loc[
        ~df["customer_age"].between(1, 100),
        "customer_age"
    ] = pd.NA

if "rating" in df.columns:
    df.loc[
        ~df["rating"].between(1, 5),
        "rating"
    ] = np.nan

# Quantity may be negative for returns, so don't delete it automatically.

# ==================================================
# 8. REMOVE EXACT DUPLICATE ROWS ONLY
# ==================================================
before_duplicates = len(df)
df = df.drop_duplicates().copy()
exact_duplicates_removed = before_duplicates - len(df)

# ==================================================
# 9. QUALITY REPORT
# ==================================================
print("\n========== CLEANING REPORT ==========")
print("Original rows:", original_rows)
print("Rows after exact duplicate removal:", len(df))
print("Exact duplicate rows removed:", exact_duplicates_removed)

print("\nMissing values per column:")
print(df.isna().sum())

print("\nMissing-value percentages:")
print((df.isna().mean() * 100).round(2))

print("\nFinal data types:")
print(df.dtypes)

print("\nFirst five rows:")
display(df.head())

# ==================================================
# 10. SAVE A NEW FILE
# ==================================================
output_path = Path.cwd() / "SWYNEX_Cleaned_Data.csv"

# Export missing values as the literal text NA
df.to_csv(output_path, index=False, na_rep="NA")

print("\nSaved cleaned file to:", output_path)
```

```python

import pandas as pd
import re

def clean_text(value):
    if pd.isna(value):
        return pd.NA

    value = str(value).strip().lower()

    # Remove special characters and extra spaces
    value = re.sub(r"[^a-z\s]", " ", value)
    value = re.sub(r"\s+", " ", value).strip()

    return value if value else pd.NA

df["product_category"] = df["product_category"].apply(clean_text)

# Standardize category names
category_mapping = {
    "body care": "Body Care",
    "bodycare": "Body Care",
    "haircare": "Haircare",
    "fragrance": "Fragrance",
    "makeup": "Makeup",
    "skincare": "Skincare",
    "skin care": "Skincare",
    "unknown": pd.NA,
    "name": pd.NA
}

df["product_category"] = df["product_category"].replace(category_mapping)

print(df["product_category"].value_counts(dropna=False))
```

```python

import pandas as pd
import re

def normalize_label(value):
    if pd.isna(value):
        return pd.NA

    value = str(value).strip().lower()
    value = re.sub(r"[^a-z\s]", " ", value)
    value = re.sub(r"\s+", " ", value).strip()

    return value if value else pd.NA

df["sales_channel"] = df["sales_channel"].apply(normalize_label)

channel_mapping = {
    "online store": "Online Store",
    "onlinestore": "Online Store",
    "retail store": "Retail Store",
    "retailstore": "Retail Store",
    "marketplace": "Marketplace",
    "social commerce": "Social Commerce",
    "socialcommerce": "Social Commerce",
    "na": pd.NA,
    "unknown": pd.NA
}

df["sales_channel"] = df["sales_channel"].replace(channel_mapping)

print(df["sales_channel"].value_counts(dropna=False))
```

```python

import pandas as pd
import numpy as np
import re
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

def normalize_text(value):
    if pd.isna(value):
        return pd.NA

    value = str(value).strip().lower()
    value = re.sub(r"[^a-z0-9\s]", " ", value)
    value = re.sub(r"\s+", " ", value).strip()

    if value in ["", "na", "n a", "n a", "null",
                 "none", "unknown", "not applicable"]:
        return pd.NA

    return value

# Standardize these columns
text_columns = [
    "product_name", "brand", "skin_type", "region",
    "stock_status", "payment_method", "returned"
]

for col in text_columns:
    df[col] = df[col].apply(normalize_text)

# Standardize category labels
mappings = {
    "product_name": {
        "body wash": "Body Wash",
        "hair growth serum": "Hair Growth Serum",
        "eau de parfum": "Eau De Parfum",
        "roll on deodorant": "Roll-On Deodorant",
        "sunscreen spf 50": "Sunscreen SPF 50"
    },
    "skin_type": {
        "not applicable": pd.NA,
        "notapplicable": pd.NA
    },
    "stock_status": {
        "in stock": "In Stock",
        "instock": "In Stock",
        "low stock": "Low Stock",
        "lowstock": "Low Stock",
        "out of stock": "Out of Stock",
        "outofstock": "Out of Stock"
    },
    "payment_method": {
        "upi": "UPI",
        "cash on delivery": "Cash on Delivery",
        "debit card": "Debit Card",
        "credit card": "Credit Card",
        "net banking": "Net Banking",
        "wallet": "Wallet"
    },
    "returned": {
        "yes": "Yes",
        "no": "No"
    },
    "region": {
        "north": "North",
        "south": "South",
        "east": "East",
        "west": "West",
        "central": "Central"
    },
    "skin_type": {
        "normal": "Normal",
        "oily": "Oily",
        "dry": "Dry",
        "sensitive": "Sensitive",
        "combination": "Combination"
    }
}

for col, mapping in mappings.items():
    df[col] = df[col].replace(mapping)

# Convert text labels to title case, preserving selected acronyms
for col in text_columns:
    df[col] = df[col].apply(
        lambda x: x.title() if isinstance(x, str) else x
    )

# Keep Yes/No capitalization consistent
df["returned"] = df["returned"].replace({
    "Yes": "Yes", "No": "No"
})

print("Categorical cleaning complete.")
display(df[text_columns].head())
```

```python

# Convert dates
df["order_date"] = pd.to_datetime(
    df["order_date"],
    errors="coerce",
    dayfirst=True
)

# Convert numeric columns
numeric_columns = [
    "unit_price_inr",
    "discount_percent",
    "quantity",
    "customer_age",
    "rating"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

print(df.dtypes)
```

```python

print("Negative prices:")
display(df[df["unit_price_inr"] < 0])

print("Discounts outside 0–100:")
display(
    df[
        (df["discount_percent"] < 0) |
        (df["discount_percent"] > 100)
    ]
)

print("Non-positive quantities:")
display(df[df["quantity"] <= 0])

print("Customer ages outside 1–100:")
display(
    df[
        (df["customer_age"] < 1) |
        (df["customer_age"] > 100)
    ]
)

print("Ratings outside 1–5:")
display(
    df[
        (df["rating"] < 1) |
        (df["rating"] > 5)
    ]
)
```

```python

df["estimated_sales_inr"] = (
    df["unit_price_inr"]
    * df["quantity"]
    * (1 - df["discount_percent"] / 100)
)

print("Total rows:", len(df))
print("Unique transaction IDs:", df["transaction_id"].nunique())
print("Average unit price (INR):", round(df["unit_price_inr"].mean(), 2))
print("Average discount (%):", round(df["discount_percent"].mean(), 2))
print("Average rating:", round(df["rating"].mean(), 2))
print("Total quantity:", df["quantity"].sum())
print(
    "Estimated sales (INR):",
    round(df["estimated_sales_inr"].sum(), 2)
)
```

```python

import pandas as pd
import numpy as np

eda_df = df.copy()

# Standardize text columns
text_cols = eda_df.select_dtypes(include=["object", "str"]).columns

for col in text_cols:
    eda_df[col] = (
        eda_df[col]
        .astype("string")
        .str.strip()
        .str.replace(r"\s+", " ", regex=True)
    )

# Standardize product categories
category_map = {
    "body-care": "Body Care",
    "bodycare": "Body Care",
    "haircare": "Haircare",
    "fragrance!": "Fragrance",
    "skincare!": "Skincare",
    "makeup*": "Makeup",
}

eda_df["product_category"] = (
    eda_df["product_category"]
    .str.lower()
    .replace(category_map)
    .str.title()
)

# Standardize sales channels
channel_map = {
    "online store": "Online Store",
    "retail store": "Retail Store",
    "social commerce": "Social Commerce",
    "marketplace": "Marketplace",
}

eda_df["sales_channel"] = (
    eda_df["sales_channel"]
    .str.lower()
    .replace(channel_map)
    .str.title()
)

# Convert placeholders into missing values
missing_values = [
    "", "na", "n/a", "nan", "none", "null",
    "not applicable", "unknown", "-", "#name?"
]

for col in text_cols:
    eda_df[col] = eda_df[col].replace(
        missing_values, pd.NA, regex=False
    )
```
